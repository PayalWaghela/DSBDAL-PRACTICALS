```python
import pandas as pd

import seaborn as sns

import matplotlib.pyplot as plt
```

```python
df1=df.drop(['PassengerId', 'Pclass', 'Name', 'SibSp',
       'Parch', 'Ticket', 'Cabin', 'Embarked'],axis=1)
```

```python
print(df1)
```

```python
sns.set_theme(style="ticks",color_codes=True)
```

```python
sns.countplot(x='Sex',data=df1)
```

```python
sns.countplot(x='Survived',data=df1)
```

```python
sns.displot(df1['Fare'],kde=False)
```

```python
df1[['Fare','Survived']]
```

```python
df1.boxplot()
```

```python
sns.histplot(df1['Fare'],bins=30,kde=False)
plt.title("Fare Distribution of Titanic Passengers")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.boxplot(x='Sex',y='Age',data=df1)
```

```python
sns.boxplot(x=df1['Sex'],y=df1['Age'],hue=df1['Survived'],palette='Set2')
```

```python
sns.boxplot(x='Survived', y='Fare', data=df1)
```
