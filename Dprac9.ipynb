```python
import pandas as pd

import seaborn as sns

import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("titanic.csv")
print(df)
```

```python
df.dropna(subset=['Age','Sex','Survived'],inplace=True)
```

```python
sns.set_style("whitegrid")
```

```python
plt.Figure(figsize=(8,6))
sns.boxplot(x='Sex',y='Age',hue='Survived',data=df,palette='Set2')
plt.title("Distribution of Passenger Ages by Gender and survival status")
plt.xlabel('Sex')
plt.ylabel('Age')
plt.show()
```
